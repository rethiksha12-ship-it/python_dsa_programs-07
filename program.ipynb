```python
def reverse_vowels(s):
    vowels = "aeiouAEIOU"
    chars = list(s)
    left, right = 0,len(chars) - 1

    while left < right:
        if chars[left] not in vowels:
            left += 1
        elif chars[right] not in vowels:
            right -= 1
    return "".join(chars)
print(reversed,("hello"))

```

```python
def remove_duplicate(arr):
    result = []
    seen = set()

    for x in arr:
        if x not in seen:
            result.append(x)
            seen.add(x)

    return result

print(remove_duplicate([1,2,2,3,1,4,3]))

```

```python
def move_zeros(arr):
    pos = 0

    for i in range(len(arr)):
        if arr[i] !=0:
            arr[pos], arr[i] = arr[i], arr[pos]
            pos += 1

    return arr

arr = [0, 1, 0, 3, 12]
print(move_zeros(arr)) 
```

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n-1)+fibonacci(n-2)


n = 7
print("fibonacci:", fibonacci(n))
```

```python
def fibonacci(n, memo={}):
    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2)
    return memo[n]

n = 7
print("fibonacci:", fibonacci(n))
```

```python
def steps_to_zero(n):
    steps = 0

    while n > 0:
        if n % 2 == 0:
            n = n // 2
        else:
            n = n - 1

        steps += 1

    return steps

n = 14
print("steps:", steps_to_zero(n))
```
