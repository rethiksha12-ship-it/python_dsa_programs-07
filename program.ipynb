```python
def reverse_vowels(s):
    vowels = "aeiouAEIOU"
    chars = list(s)
    left, right = 0,len(chars) - 1

    while left < right:
        if chars[left] not in vowels:
            left += 1
        elif chars[right] not in vowels:
            right -= 1
    return "".join(chars)
print(reversed,("hello"))

```

```python
def remove_duplicate(arr):
    result = []
    seen = set()

    for x in arr:
        if x not in seen:
            result.append(x)
            seen.add(x)

    return result

print(remove_duplicate([1,2,2,3,1,4,3]))

```

```python
def second_largest(arr):
    largest=second=float("-inf")
    second = float('-inf')
    for num in arr:
        if num > largest:
            second = largest
            largest = num 
        elif num > second and num != largest:
            second = num

    return second
arr = [10,5,8,5,1]
print("second largest:", second_largest(arr))

```

```python
def move_zeros(arr):
    pos = 0

    for i in range(len(arr)):
        if arr[i] !=0:
            arr[pos], arr[i] = arr[i], arr[pos]
            pos += 1

    return arr

arr = [0, 1, 0, 3, 12]
print(move_zeros(arr)) 
```

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n-1)+fibonacci(n-2)


n = 7
print("fibonacci:", fibonacci(n))
```

```python
def fibonacci(n, memo={}):
    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = fibonacci(n - 1, memo) + fibonacci(n - 2)
    return memo[n]

n = 7
print("fibonacci:", fibonacci(n))
```

```python
def steps_to_zero(n):
    steps = 0

    while n > 0:
        if n % 2 == 0:
            n = n // 2
        else:
            n = n - 1

        steps += 1

    return steps

n = 14
print("steps:", steps_to_zero(n))
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def rotate(head, k):
    if head is None or head.next is None:
        return head

    # Find length and last node
    temp = head
    length = 1

    while temp.next:
        temp = temp.next
        length += 1

    k = k % length
    if k == 0:
        return head

    # Make circular
    temp.next = head

    # Find new last node
    for i in range(length - k):
        temp = temp.next

    new_head = temp.next
    temp.next = None

    return new_head

def display(head):
    while head:
        print(head.data, end=" -> ")
        head = head.next
    print("None")

# Create linked list
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)
head.next.next.next.next = Node(50)

print("Original List:")
display(head)

head = rotate(head, 2)

print("After Rotation:")
display(head)
```

```python
from collections import deque

d = deque()

# Insert at rear
d.append(10)
d.append(20)

# Insert at front
d.appendleft(5)

print("Deque:", d)

# Delete from front
d.popleft()

# Delete from rear
d.pop()

print("After deletion:", d)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def postorder(root):
        if root is not None:
            postorder(root.left)
            print(root.data, end=" ")
            postorder(root.right)

root = Node(1)

root.left = Node(2)
root.right = Node(3)

root.left.left = Node(4)
root.left.right = Node(5)

root.right.left = Node(6)
root.right.right = Node(7)


print("postorder traversal:")
postorder(root)
```
